```python
import pandas as pd

data = pd.read_csv("1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")

data.head()
```

```python
print("Fuel Type:")
print(data['Fuel_Type'].value_counts())

print("\nSeller Type:")
print(data['Seller_Type'].value_counts())

print("\nTransmission:")
print(data['Transmission'].value_counts())

print("\nOwner:")
print(data['Owner'].value_counts())
```

```python
print("Oldest Manufacturing Year:", data['Year'].min())
print("Latest Manufacturing Year:", data['Year'].max())
```

```python
data['Year'].value_counts().sort_index()
```

```python
print("Lowest Selling Price:", data['Selling_Price'].min(), "lakh")
print("Highest Selling Price:", data['Selling_Price'].max(), "lakh")
print("Average Selling Price:", round(data['Selling_Price'].mean(), 2), "lakh")
```

```python
data.loc[data['Selling_Price'].idxmax()]
```

```python
data[['Car_Name', 'Year', 'Selling_Price']].sort_values(
    by='Selling_Price',
    ascending=False
).head(10)
```

```python
data[['Car_Name', 'Year', 'Selling_Price']].sort_values(
    by='Selling_Price',
    ascending=True
).head(10)
```

```python
fuel_counts = data['Fuel_Type'].value_counts()

print(fuel_counts)
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    fuel_counts,
    labels=fuel_counts.index,
    autopct='%1.1f%%'
)

plt.title("Distribution of Cars by Fuel Type")
plt.show()
```

```python
seller_counts = data['Seller_Type'].value_counts()

print(seller_counts)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(seller_counts.index, seller_counts.values)

plt.xlabel("Seller Type")
plt.ylabel("Number of Cars")
plt.title("Cars by Seller Type")

plt.show()
```

```python
transmission_counts = data['Transmission'].value_counts()

print(transmission_counts)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(transmission_counts.index, transmission_counts.values)

plt.xlabel("Transmission Type")
plt.ylabel("Number of Cars")
plt.title("Manual vs Automatic Cars")

plt.show()
```

```python
owner_counts = data['Owner'].value_counts().sort_index()

print(owner_counts)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(owner_counts.index.astype(str), owner_counts.values)

plt.xlabel("Number of Previous Owners")
plt.ylabel("Number of Cars")
plt.title("Cars by Number of Previous Owners")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(data['Kms_Driven'], data['Selling_Price'])

plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Selling Price vs Kilometres Driven")

plt.show()
```

```python
avg_fuel_price = data.groupby('Fuel_Type')['Selling_Price'].mean()

print(avg_fuel_price)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(avg_fuel_price.index, avg_fuel_price.values)

plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price (Lakhs)")
plt.title("Average Selling Price by Fuel Type")

plt.show()
```

```python
avg_transmission_price = data.groupby('Transmission')['Selling_Price'].mean()

print(avg_transmission_price)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(avg_transmission_price.index, avg_transmission_price.values)

plt.xlabel("Transmission Type")
plt.ylabel("Average Selling Price (Lakhs)")
plt.title("Average Selling Price by Transmission")

plt.show()
```

```python
avg_year_price = data.groupby('Year')['Selling_Price'].mean()

print(avg_year_price)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(avg_year_price.index, avg_year_price.values, marker='o')

plt.xlabel("Manufacturing Year")
plt.ylabel("Average Selling Price (Lakhs)")
plt.title("Average Selling Price by Manufacturing Year")

plt.xticks(rotation=45)

plt.show()
```

```python
top_10_expensive = data[
    ['Car_Name', 'Year', 'Selling_Price', 'Present_Price']
].sort_values(
    by='Selling_Price',
    ascending=False
).head(10)

top_10_expensive
```

```python
top_depreciation = data[
    ['Car_Name', 'Present_Price', 'Selling_Price', 'Depreciation']
].sort_values(
    by='Depreciation',
    ascending=False
).head(10)

top_depreciation
```
